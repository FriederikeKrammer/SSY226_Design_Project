```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Read the CSV file
df = pd.read_csv("data.csv")

# 2. Convert the time column to datetime format
df["Time"] = pd.to_datetime(df["Time"])

# 3. Set time as the index (optional)
df.set_index("Time", inplace=True)

# 4. Display the first few rows
print(df.head())

# 5. Plot the time-series data
plt.figure(figsize=(12, 6))

plt.plot(df.index, df["Value"], 
         label="Time Series", 
         linewidth=1.5)

# 6. Add plot labels and title
plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Time-Series Data")

plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

# 7. Adjust layout and display
plt.tight_layout()
plt.show()
```
