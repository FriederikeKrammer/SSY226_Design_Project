```python

import pandas as pd
import plotly.graph_objects as go

# ======================================
# CSV einlesen
# ======================================

df = pd.read_csv("data_HL-T-sun_2021-2025.csv")

df["datetime"] = pd.to_datetime(df["datetime"])

# ======================================
# Plot erstellen
# ======================================

fig = go.Figure()

# Verbrauch - linke Y-Achse
fig.add_trace(
    go.Scatter(
        x=df["datetime"],
        y=df["cons"],
        name="Consumption",
        mode="lines",
        yaxis="y"
    )
)

# Temperatur - rechte Y-Achse
fig.add_trace(
    go.Scatter(
        x=df["datetime"],
        y=df["temperature"],
        name="Temperature",
        mode="lines",
        yaxis="y2"
    )
)

# Sonne - zweite rechte Y-Achse
fig.add_trace(
    go.Scatter(
        x=df["datetime"],
        y=df["sun"],
        name="Solar Radiation",
        mode="lines",
        yaxis="y3"
    )
)

# ======================================
# Layout
# ======================================

fig.update_layout(

    title="Consumption, Temperature and Solar Radiation (2021-2025)",

    xaxis=dict(
        title="Date",
        domain=[0.10, 0.82]
    ),

    yaxis=dict(
        title="Consumption",
        side="left"
    ),

    yaxis2=dict(
        title="Temperature [°C]",
        overlaying="y",
        side="right"
    ),

    yaxis3=dict(
        title="Solar Radiation",
        overlaying="y",
        side="right",
        anchor="free",
        position=0.90
    ),

    template="plotly_white",

    hovermode="x unified",

    height=750,

    legend=dict(
        orientation="h",
        y=1.12,
        x=0
    ),

    margin=dict(
        l=90,
        r=160,
        t=120,
        b=80
    )
)

# ======================================
# Anzeige und Export
# ======================================



fig.write_html("cons_temperature_sun.html")

import webbrowser
webbrowser.open("cons_temperature_sun.html")
```
